# Stemming 
## doesnt have a dictionary in it

stemming gives stem

```python
from nltk.stem import PorterStemmer
words=['eat','eaten','broke','break','broken','bride','class','classify','nation','probation']
stemmer=PorterStemmer()
for word in words:
    print("word is --->", stemmer.stem(word))
```

## Notice how many words are wrongly stemmed

```python
#regexstemming
from nltk.stem import RegexpStemmer
stemmer=RegexpStemmer('ing$|ed$', min=4)
for word in words:
    print("words are --->", stemmer.stem(word))
```

```python
#snowballstemming
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer('english')
for word in words:
    print(snowball_stemmer.stem(word))
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

steps:
1. tokenize
2. stop words
3. stemming

```python
stemmer=PorterStemmer()
```

```python
sentences=nltk.sent_tokenize(paragraph)
type(sentences)
```

```python
## Apply Stopwords And Filter And then Apply Stemming

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
```
