# Stemming 
## doesnt have a dictionary in it

stemming gives stem

```python
from nltk.stem import PorterStemmer
words=['eat','eaten','broke','break','broken','bride','class','classify','nation','probation']
stemmer=PorterStemmer()
for word in words:
    print("word is --->", stemmer.stem(word))
```

## Notice how many words are wrongly stemmed

```python
#regexstemming
from nltk.stem import RegexpStemmer
stemmer=RegexpStemmer('ing$|ed$', min=4)
for word in words:
    print("words are --->", stemmer.stem(word))
```

```python
#snowballstemming
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer('english')
for word in words:
    print(snowball_stemmer.stem(word))
```

```python
import nltk
nltk.download('wordnet')
```

```python
words=['eat','eaten','broke','break','broken','bride','class','classify','nation','probation']
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
for word in words:
    print(word+ "--->",lemmatizer.lemmatize(word))
```
