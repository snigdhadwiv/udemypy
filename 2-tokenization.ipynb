```python

corpus=""" Hi my name is basanti. 
I eat Chawal basmati ! 
I can poop in my pants."""
print(corpus)
```

```python
# tokenization
# paragraph into a sentence
from nltk.tokenize import sent_tokenize
sentences=sent_tokenize(corpus)
sentences
```

```python
import nltk

nltk.download('punkt_tab')
```

```python
for sentence in sentences: 
    print(sentence)
```

```python
#corpus tokenized into words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in sentences: 
    print(word_tokenize(sentence))
```

```python
#to tokenize punchuations as well, we use wordpunct_tokenize
#for example
punct_corpus="""I like Mira's eyes, They are Beautiful. She is A little crazy though. I like her alot."""
punct_corpus
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(punct_corpus)
#check the Mira's part, it should be spearated as well.
```

```python
from nltk.tokenize import TreebankWordDetokenizer
treebank=TreebankWordDetokenizer()
treebank.tokenize(punct_corpus)
```
