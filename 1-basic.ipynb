```python
#simple calculator
num1=int(input("enter first number"))
num2=int(input("enter second number"))
operation=str(input("enter operation"))

if operation=="+":
    print(num1+num2)
elif operation=="-":
    print(num1-num2)
elif operation=="/":
    if num2!=0:
        print(num1/num2)
    else:
        print(num2,"is zero, division by zero is incorrect.") 
elif operation=="*":
    print(num1*num2)
elif operation=="%":
    print(num1%num2)
elif operation=="^":
    print(num1**num2)
else: 
    print("incorrect operation")
```

```python
#sum of first 10 natural numbers
#iterative sum (loop sum(loop type : while))
n=10
sum=0
count=0
while count<=n:
    sum=sum+count
    count=count+1

print("sum of first n natural numbers", sum)
```
